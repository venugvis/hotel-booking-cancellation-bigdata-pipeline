# Tableau Storyboarding: Hotel Booking Cancellation

In this notebook, I’m preparing the final dataset and visual design ideas for my Tableau storyboard.

The objective is to create a compelling visual narrative that:
- Highlights the most important factors driving cancellations.
- Shows patterns across hotel types, months, customer segments, and more.
- Displays the output of the Random Forest model to explain predictive insights.

This sets the stage for communicating findings effectively with business stakeholders.


```python
import pandas as pd

# Reload cleaned full dataset
df = pd.read_csv('../data/processed/hotel_bookings_cleaned.csv')

# Include important categorical columns for Tableau
export_cols = [
    'is_canceled', 'hotel', 'arrival_date_month', 'lead_time', 'adr',
    'market_segment', 'customer_type', 'deposit_type', 'total_of_special_requests',
    'booking_changes', 'arrival_month_num', 'days_in_waiting_list',
    'is_repeated_guest', 'required_car_parking_spaces', 'previous_cancellations'
]

# Create export-ready DataFrame
tableau_df = df[export_cols]

# Save for Tableau
tableau_df.to_csv('../data/processed/hotel_cancellation_for_tableau.csv', index=False)
```

## Storyboard Layout in Tableau

Here’s a rough layout of the visual story I plan to build:

1. **Overview Metrics**
   - Cancellation rate
   - Hotel type distribution

2. **Temporal Trends**
   - Cancellations by month
   - Lead time vs. cancellations

3. **Customer Behavior**
   - Market segment vs cancellation rate
   - Repeated guests vs first-time guests

4. **Operational Factors**
   - Deposit type impact
   - Special requests, booking changes

5. **Model Insight**
   - Feature importance from Random Forest
   - Summary model performance (optional static image)

The dashboard will be built in Tableau Public and linked back to this notebook or GitHub repository.
